```python
s1 = "Tokenization is essential in NLP."
```

```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp(s1)
```

```python
for token in doc:
  print(token)
```

```python
doc2 = doc[1:len(s1)]
```

```python
print(doc2)
```
