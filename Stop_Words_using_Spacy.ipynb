```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
c = nlp.Defaults.stop_words
```

```python
len(nlp.Defaults.stop_words)
```

```python
nlp.vocab['better'].is_stop
```

```python
s1 = "This is an example sentence with some stop words that we want to remove."
```

```python
doc = nlp(s1)
```

```python
token_without_stopwords = [token.text for token in doc if token.text.lower() not in nlp.Defaults.stop_words]
```

```python
token_without_stopwords
```

```python
# add our own stopwords into the spacy stopwords library
nlp.Defaults.stop_words.add("example")
```

```python
token_without_stopwords = [token.text for token in doc if token.text.lower() not in nlp.Defaults.stop_words]
```

```python
token_without_stopwords
```

```python
# remove our own stopwords into the spacy stopwords library
nlp.Defaults.stop_words.remove("example")
```

```python
token_without_stopwords = [token.text for token in doc if token.text.lower() not in nlp.Defaults.stop_words]
```

```python
token_without_stopwords
```

```python
# in sentiment analysis, stop_words can give inaccurate results. Since 'not' contains in stop_words, it can create confusion by removing not from a negative review. can cause issues with sentiment analysis
s2 = "this is not a good course"
```

```python
token_without_stopwords = [token.text for token in nlp(s2) if token.text.lower() not in nlp.Defaults.stop_words]
```

```python
token_without_stopwords
# output: ['good', 'course'] which is wrong in case of sentiment analysis
```
